**Step 1 — Import Pandas**

Pandas is a Python library mainly used for data manipulation and analysis.

````python
import pandas as pd  # Import Pandas library
````

**Why do we use Pandas?**

Pandas provides two important data structures:

**Series** → One-dimensional data

**DataFrame** → Two-dimensional tabular data

A DataFrame looks similar to an Excel table.

**Step 2 — Create the Dataset**

Let's create a dataset containing 15 customer orders.

````python
# Create customer order data using a dictionary
data = {
    "Order_ID": [1001, 1002, 1003, 1004, 1005,
                 1006, 1007, 1008, 1009, 1010,
                 1011, 1012, 1013, 1014, 1015],

    "Customer": ["Aman", "Priya", "Rahul", "Simran", "Karan",
                 "Neha", "Arjun", "Riya", "Vikas", "Anjali",
                 "Mohit", "Pooja", "Rohit", "Sneha", "Varun"],

    "City": ["Delhi", "Mumbai", "Chandigarh", "Pune", "Delhi",
             "Jaipur", "Mumbai", "Chandigarh", "Bangalore", "Delhi",
             "Pune", "Mumbai", "Jaipur", "Delhi", "Bangalore"],

    "Product": ["Laptop", "Mobile", "Headphones", "Tablet", "Keyboard",
                "Monitor", "Mobile", "Mouse", "Laptop", "Tablet",
                "Headphones", "Laptop", "Mobile", "Monitor", "Keyboard"],

    "Category": ["Electronics", "Electronics", "Accessories", "Electronics",
                 "Accessories", "Electronics", "Electronics", "Accessories",
                 "Electronics", "Electronics", "Accessories", "Electronics",
                 "Electronics", "Electronics", "Accessories"],

    "Quantity": [1, 2, 1, 1, 2, 1, 1, 3, 1, 2, 2, 1, 2, 1, 2],

    "Price": [55000, 25000, 2000, 30000, 1500,
              18000, 27000, 800, 60000, 32000,
              2500, 58000, 24000, 20000, 1800],

    "Payment": ["UPI", "Card", "Cash", "UPI", "Card",
                "UPI", "Card", "Cash", "UPI", "Card",
                "UPI", "Card", "Cash", "UPI", "Card"],

    "Status": ["Delivered", "Delivered", "Pending", "Shipped", "Delivered",
               "Delivered", "Pending", "Delivered", "Shipped", "Delivered",
               "Cancelled", "Delivered", "Pending", "Shipped", "Delivered"]
}

# Convert the dictionary into a Pandas DataFrame
df = pd.DataFrame(data)

# Display the DataFrame
df
````

````python
# Display only the DataFrame index
df.index
````

Important distinction

`Index: 0, 1, 2, 3, ...`

`Order_ID: 1001, 1002, 1003, ...`

Order_ID is a column, while the default Pandas index is used to identify rows.

**Step 5 — Selecting a Single Column**

````python
# Select the Customer column
print(df["Customer"])
print("\n\nThe result is a Pandas Series.")
````

````python
# Select the Customer column using dot notation
df.Customer
````

This works in many cases, but the recommended approach is:

`df["Customer"]`

because it works reliably with column names containing spaces or special characters.

**Step 6 — Selecting Multiple Columns**

````python
# Select multiple columns
df[["Customer", "Product", "Price"]]
````

Why double brackets?

Outer brackets:

`df[ ... ]`

tell Pandas that we are selecting data.

Inner brackets:

`["Customer", "Product", "Price"]`

contain the list of columns.

**Step 7 — Selecting Rows Using Indexing**

We can select rows based on their index.

**Select one row**

````python
# Select the row with index 3
df.iloc[3]
````

**Step 8 — .iloc[] Indexing**

`.iloc[]` means integer-location based indexing.

It selects data using numerical positions.

**Select first row**

````python
# Select the first row using its integer position
df.iloc[0]
````

````python
# Select the fifth row
df.iloc[4]
````

````python
# Select rows at positions 0, 2 and 4
df.iloc[[0, 2, 4]]
````

**Step 9 — Row Slicing Using .iloc[]**

````python
# Select rows from position 1 to position 4, Excluded 5th position
df.iloc[1:5]
````

**Step 10 — Selecting Specific Rows and Columns Using .iloc[]**

Synatx:
`df.iloc[row_selection, column_selection]`

````python
# Select rows 0 to 4 and columns 1 to 3
df.iloc[0:5, 1:4]
````

````python
# Select specific rows and specific columns
# Select rows 0, 2 and 4 and columns 1 and 3
df.iloc[[0, 2, 4], [1, 3]]
````

**Step 11 — .loc[] Indexing**

.loc[] is label-based indexing.

Syntax:

`df.loc[row_label, column_label]`

````python
# Select the row having index label 3
df.loc[3]
````

````python
# Select the customer from row label 3
df.loc[3, "Customer"]
````

**Step 12 — .loc[] for Multiple Columns**

````python
# Select Customer, City and Product for row 3
df.loc[3, ["Customer", "City", "Product"]]
````

**Step 13 — .loc[] for Multiple Rows**

````python
# Select rows from index 2 through index 5
df.loc[2:5]
````

**Important difference**

**With .loc[]:**

`df.loc[2:5]`

includes 5.

**With .iloc[]:**

`df.iloc[2:5]`

does not include position 5.

This is an important exam/interview concept.

````python
df.loc[2:5] # includes 5
````

````python
df.iloc[2:5] # Not include 5
````

**Step 14 — .loc[] for Rows and Columns**

````python
# Select rows 2 to 5 and only Customer and Product columns
df.loc[2:5, ["Customer", "Product"]]
````

**Step 15 — Conditional Selection**

Now we move to a very important real-world operation: **filtering**.


Suppose we want customers who purchased from Delhi.

````python
# Select orders where the city is Delhi
df[df["City"] == "Delhi"]
````

**Step 16 — Filter Orders Above ₹30,000**

````python
# Select orders where the price is greater than 30000
df[df["Price"] > 30000]
````

**Step 17 — Filter Based on Multiple Conditions**

Suppose we want:

>Electronics products costing more than ₹30,000.



````python
# Select electronics orders costing more than 30000
df[(df["Category"] == "Electronics") & (df["Price"] > 30000)]
````

**Important**

For multiple conditions, use: &
 for AND

and: | for OR

Do not use Python's normal:
and,
or

with Pandas Series.

**Step 18 — OR Condition**

Suppose we want orders from either Delhi or Mumbai.

````python
# Select orders from Delhi or Mumbai
df[(df["City"] == "Delhi") | (df["City"] == "Mumbai")]
````

**Step 19 — Filtering Using .loc[]**

We can combine .loc[] with conditions.

````python
# Select Customer and Product for orders from Delhi
df.loc[df["City"] == "Delhi", ["Customer", "Product"]]
````

This is particularly useful because we can **filter rows and select columns simultaneously.**

**Step 20 — Select Delivered Orders**

````python
# Select all orders that have been delivered
df[df["Status"] == "Delivered"]
````

**Step 21 — Select Pending Orders**

````python
# Select orders that are still pending
df[df["Status"] == "Pending"]
````

**Step 22 — Select Customers Who Bought Laptops**

````python
# Select orders where the purchased product is a Laptop
df[df["Product"] == "Laptop"]
````

**Step 23 — Changing a Value Using .loc[]**

Pandas also allows us to modify values.

Suppose order 1003 has been shipped.

````python
# Change the status of order 1003 to Shipped
df.loc[df["Order_ID"] == 1003, "Status"] = "Shipped"
````

````python
# Display the updated order
df[df["Order_ID"] == 1003]
````

**Step 24 — Adding a New Column**

We can calculate the total order value.

Formula:

`Total Amount = Quantity × Price`

````python
# Calculate total amount for each order
df["Total_Amount"] = df["Quantity"] * df["Price"]

# Display the updated DataFrame
df
````

**Step 25 — Select Orders Based on Total Amount**

Now that we have calculated Total_Amount, we can use it for filtering.

````python
# Select orders with total value greater than 50000
df[df["Total_Amount"] > 50000]
````

**Step 26 — Remove a Column**

Suppose we want to remove the Payment column.

````python
# Remove the Payment column
df = df.drop("Payment", axis=1)

# Display the updated DataFrame
df
````

Here:

`axis=1`

means we are removing a column.

For rows:

`axis=0`

**Step 27 — Reset the Index**

Sometimes after filtering, we may want to create a fresh sequential index.

````python
# Filter delivered orders
delivered_df = df[df["Status"] == "Delivered"]

# Reset the index after filtering
delivered_df = delivered_df.reset_index(drop=True)

# Display the result
delivered_df
````

**Step 28 — A Practical Example**

Suppose the shopping platform asks:

>Find customers from Mumbai who purchased products costing more than ₹20,000.

We can solve it using:

````python
# Filter Mumbai customers with products costing more than 20000
result = df.loc[
    (df["City"] == "Mumbai") & (df["Price"] > 20000),
    ["Customer", "Product", "Price", "Status"]
]

# Display the selected information
result
````

This combines:

```
.loc[]
Boolean conditions
&
Multiple-column selection
```



This is a very realistic DataFrame operation.
