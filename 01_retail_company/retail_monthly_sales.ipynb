```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df["City"] = df["City"].fillna("unknown")
```

```python
df["Customer_Rating"] = df["Customer_Rating"].fillna(df["Customer_Rating"].mean())
```

```python
df[df.isnull().any(axis = 1)]
```

```python
df[df["Quantity"] < 0]
```

```python
df = df[df["Quantity"] > 0]
df
```

```python
df[df["Quantity"] < 0]
```

```python
df["City"]
```

```python
df["City"] = df["City"].str.title()
df["Product"] = df["Product"].str.title()
df["Gender"] = df["Gender"].str.title()
df["Order_Status"] = df["Order_Status"].str.title()
```

```python
df.info()
```

```python
df["Order_Date"] = pd.to_datetime(df["Order_Date"])
```

```python
df["Month"] = df["Order_Date"].dt.month_name()
df["Year"] = df["Order_Date"].dt.year
```

```python
df["Total_Sales"] = df["Quantity"] * df["Unit_Price"]
```

```python
# Basic EDA
df["Total_Sales"].sum()
```

```python
df["Total_Sales"].mean()
```

```python
df["Total_Sales"].min()
```

```python
df["Total_Sales"].max()
```

```python
# Monthly Sales
monthly_sales = df.groupby("Month")["Total_Sales"].sum()
monthly_sales.sort_values(ascending = False).head(1)
```

```python
df.groupby("Product")["Total_Sales"].sum().sort_values(ascending=False)
```

```python
df.groupby("Category")["Total_Sales"].sum()
```

```python
df.groupby("City")["Total_Sales"].sum()
```

```python
df.groupby("Salesperson")["Total_Sales"].sum()
```

```python
# Best selling Product

df.groupby("Product")["Quantity"].sum().sort_values(ascending=False)
```

```python
# payment Mode

df["Payment_Mode"].value_counts()
```

```python
df.groupby("Product")["Customer_Rating"].mean().sort_values().tail(1)
```

```python
# Visualization
# monthly sales
# monthly_sales.plot(kind = "bar", figsize = (8,4))
monthly_sales.plot(kind = "bar", figsize = (8,4), color = "orange")
plt.title("Monthly Sales")
plt.show()
```

```python
# category
category = df.groupby("Category")["Total_Sales"].sum()
category.plot(kind = 'bar', figsize=(8,4), color = "blue")
plt.show()
```

```python
# Pie
df["Category"].value_counts().plot(kind="pie", autopct = "%1.1f%%")
plt.show()
```

```python
# City
city = df.groupby("City")["Total_Sales"].sum()
city.plot(kind = "barh", figsize = (8,4), color = "red")
plt.show()
```

```python
# Histogram
plt.hist(df["Customer_Rating"])
plt.show()
```
