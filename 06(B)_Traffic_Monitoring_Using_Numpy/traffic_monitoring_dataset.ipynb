**17. Boolean Operations**

Now we reach the main objective of this problem.

Suppose the traffic department defines:

>High-density traffic = more than 1,000 vehicles.


````python
# We can create a Boolean condition:
high_traffic = vehicles > 1000
print(high_traffic)
````

The result looks like:

>[False  True False True ...]

This is a **Boolean mask.**

**18. What is Masking?**

For example:

```
vehicles > 1000
```

creates something like:

```
False
True
False
True
True
...
```

Then:

`vehicles[vehicles > 1000]`

returns only the values where the condition is **True**.

````python
# Question 17: Find all regions having more than 1,000 vehicles.
high_density = traffic[traffic[:, 1] > 1000]

print(high_density)
````

Read it as:

>Select rows from traffic where column 1 is greater than 1000.

**20. Display Only Region IDs**

**21. Count High-Density Regions**

**22. Boolean AND Operation**

Now suppose we define congestion as:

>Vehicles > 1,000 AND average speed < 30 km/h.

````python
# Use & for AND ... not and
condition = (traffic[:, 1] > 1000) & (traffic[:, 2] < 30)
congested = traffic[condition]
print(congested)
````

````python
# 23. Find IDs of Congested Regions
congested_ids = traffic[condition, 0]
print(congested_ids)
````

**24. Boolean OR Operation**

Suppose the traffic department wants to identify regions where:

> Vehicles > 1,500 OR speed < 20 km/h.

````python
# Use | for OR .. Not or
condition = (traffic[:, 1] > 1500) | (traffic[:, 2] < 20)
result = traffic[condition]
print(result)
````

**25. Boolean NOT Operation**

Suppose we want regions where traffic is not greater than 1,000.

````python
# Use ~ for NOT ... Don't use NOT
condition = ~(traffic[:, 1] > 1000)
result = traffic[condition]
print(result)
````

**30. Find the Most Congested Region**

We can use argmax().

````python
# Question 20: Find the index of the region with maximum traffic.
index = np.argmax(traffic[:, 1]) # argmax() gives the position, not the actual maximum value.
print(index)
````

````python
# Now find its details:
print("Region ID:", traffic[index, 0])
print("Vehicles:", traffic[index, 1])
print("Speed:", traffic[index, 2])
````

````python
# 31. Find Least Busy Region
index = np.argmin(traffic[:, 1])
print("Region ID:", traffic[index, 0])
print("Vehicles:", traffic[index, 1])
````

````python
# 32. Ranking Regions Using argsort()
# Suppose we want to rank regions according to traffic volume.
ranking = np.argsort(traffic[:, 1])
print(ranking)

# This returns row indices sorted according to vehicle count.
````

````python
# For descending order:
ranking = np.argsort(traffic[:, 1])[::-1]
print(ranking)
````

````python
# Now display the top 5 regions:
top5 = traffic[ranking[:5]]
print(top5)
````

This is a very useful application of **fancy indexing.**

````python
# 33. Top 5 High-Traffic Regions

vehicles = traffic[:, 1]
ranking = np.argsort(vehicles)[::-1]
top5 = traffic[ranking[:5]]
print(top5)
````

Read this as:

```
Sort vehicle positions.
Reverse them for descending order.
Take first 5 positions.
Use those positions to select rows.
```

**43. A Useful Real-World Traffic Metric**

We can create a simple congestion score.

For example:

`Congestion Score = Vehicles / Average Speed`

````python
congestion_score = traffic[:, 1] / traffic[:, 2]
print(congestion_score)
````

Higher score means:

>Many vehicles + low speed → greater congestion.

Now find the region with the highest score:

````python
index = np.argmax(congestion_score)

print("Most congested region:", traffic[index, 0])
print("Congestion score:", congestion_score[index])
````

This is a nice extension because students see how NumPy can be used to create a **new analytical metric.**

**44. Important Difference: Masking vs Fancy Indexing**

Students often confuse these two.

Masking

You select data using a condition:

`traffic[traffic[:, 1] > 1000]`

Meaning:

>Select regions where vehicles > 1000.

Fancy indexing

You select data using specific indices:

`traffic[[1, 4, 9, 13]]`

Meaning:

>Select rows 1, 4, 9 and 13.

Easy way to remember

```
Masking       → "Which data satisfies my condition?"
Fancy indexing → "Give me these specific positions."
```
