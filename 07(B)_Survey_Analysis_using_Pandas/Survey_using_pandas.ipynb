**PART A — HANDLING MISSING DATA**

**What is Missing Data?**

Missing data means that some values are unavailable.

In real-world datasets, missing data can occur because:
```
Customer skipped a question
Form was incomplete
Data was not recorded
Technical problem occurred
Information was unavailable
```



Detect Missing Values

Pandas provides:

`isnull()`

````python
# Check which values are missing
df.isnull()
````

This produces True and False.

True  → Missing

False → Not missing

**Count Missing Values**

````python
# Count missing values in each column
df.isnull().sum()
````

**Find Rows Having Missing Values**

````python
# Display rows containing at least one missing value
df[df.isnull().any(axis=1)]
````

Understanding:
`any(axis=1)`

means:

>Check each row and determine whether at least one value is missing.

**Find Complete Records**

Sometimes we want only records where no values are missing.

````python
# Select rows that contain no missing values
complete_df = df.dropna()

# Display complete records
complete_df
````

**Remove Rows with Missing Values**

dropna() removes rows containing missing values.

````python
# Remove rows containing any missing value
df_clean = df.dropna()

# Display the cleaned dataset
df_clean
````

**But there is a problem**

If the dataset has many missing values, dropna() may remove too many useful records.

Therefore, sometimes we should fill missing values instead of deleting them.

**Fill Missing Age with Mean**

````python
# Calculate the average age
mean_age = df["Age"].mean()

# Fill missing ages with the average age
df["Age"] = df["Age"].fillna(mean_age)

# Display the updated Age column
df["Age"]
````

**Fill Missing Rating with Median**

For ratings, median can be useful because it is less affected by extreme values.

````python
# Calculate the median rating
median_rating = df["Rating"].median()

# Fill missing ratings with the median
df["Rating"] = df["Rating"].fillna(median_rating)

df["Rating"]
````

**Fill Missing City**

For categorical data such as City, mean or median doesn't make sense.

We can use the **mode**.

Mode means:

>The value that occurs most frequently.

````python
# Find the most frequently occurring city
mode_city = df["City"].mode()[0]

# Fill missing cities with the most common city
df["City"] = df["City"].fillna(mode_city)

df["City"]
````

**Fill Missing Feedback**

For text fields, we can use a meaningful label.

````python
# Replace missing feedback with a descriptive label
df["Feedback"] = df["Feedback"].fillna("No feedback provided")
df["Feedback"]
````

**Fill Missing Response Channel**

````python
# Replace missing response channels with an appropriate label
df["Response_Channel"] = df["Response_Channel"].fillna("Unknown")
df["Response_Channel"]
````

**Check Missing Values Again**

````python
# Check whether any missing values remain
df.isnull().sum()
````

**PART B — CLEANING INCONSISTENT RECORDS**

Now we have to deal with another common real-world problem.

The same city has been entered in different ways:

```
Delhi
delhi

Mumbai
MUMBAI
mumbai

Pune
pune
PUNE
```

Pandas treats these as different strings.

**Remove Extra Spaces**

````python
# Remove unnecessary spaces from city names
df["City"] = df["City"].str.strip()
````

**Standardize Letter Case**

````python
# Convert city names to lowercase
df["City"] = df["City"].str.lower()
````

**Convert City Names to Title Case**

For better presentation:

````python
# Convert city names to title case
df["City"] = df["City"].str.title()
````

**Clean Response Channel**

````python
# Remove spaces and standardize response channel formatting
df["Response_Channel"] = df["Response_Channel"].str.strip().str.title()
````

**Clean Feedback Text**

We can also standardize the feedback text.

````python
# Remove extra spaces and standardize feedback capitalization
df["Feedback"] = df["Feedback"].str.strip().str.title()
````

**PART C — FINDING DUPLICATE RECORDS**

**What is Duplicate Data?**

A duplicate record is a record that appears more than once when it should appear only once.

**PART D — MERGING DATASETS**

Now let's assume the company has another dataset containing customer information.

**Load the Customer Master Dataset**

````python
# Load the customer master dataset
customer_master = pd.read_csv("customer_master.csv")

# Display the first five records
customer_master.head()
````

**Understand Why We Need Merge**

Before merging:

**Feedback dataset**

`Customer_ID | Customer_Name | Age | City | Rating`

Customer master

`Customer_ID | Email | Membership`

After merging:

`Customer_ID | Customer_Name | Age | City | Rating | Email | Membership`

This gives us a more complete dataset.

**Inner Merge**

The basic syntax is:

`pd.merge(left, right, on="common_column", how='inner')`

````python
# Merge feedback data with customer master data using Customer_ID
merged_df = pd.merge(
    df,
    customer_master,
    on="Customer_ID",
    how="inner"
)

# Display the merged dataset
merged_df.head()
````

31. Understanding inner

An inner join returns only records that have matching IDs in both datasets.

Conceptually:

```
Feedback Dataset
       +
Customer Master
       ↓
Matching Customer_IDs
```

**Left Merge**

A left merge keeps all records from the left DataFrame.

````python
# Keep all feedback records and add matching customer information
left_merged = pd.merge(
    df,
    customer_master,
    on="Customer_ID",
    how="left"
)

# Display the result
left_merged.head()
````

This is often useful when the feedback dataset is your **main dataset.**

**Right Merge**

A right merge keeps all records from the right DataFrame.

````python
# Keep all customer master records during the merge
right_merged = pd.merge(
    df,
    customer_master,
    on="Customer_ID",
    how="right"
)

# Display the result
right_merged.head()
````

**Outer Merge**

An outer merge keeps records from both datasets.

````python
# Keep all records from both datasets
outer_merged = pd.merge(
    df,
    customer_master,
    on="Customer_ID",
    how="outer"
)

# Display the merged result
outer_merged.head()
````

**Understand the Four Merge Types**

```
Merge	Meaning
inner	Only matching records
left	All records from left DataFrame
right	All records from right DataFrame
outer	All records from both DataFrames
```

Easy memory trick:

```
INNER → Intersection
LEFT  → Everything from LEFT
RIGHT → Everything from RIGHT
OUTER → Everything from BOTH
```

**PART E — CONCATENATING DATASETS**

**Load Batch 2**

````python
# Load the second feedback batch
feedback_batch2 = pd.read_csv("customer_feedback_batch2.csv")

# Display the second batch
feedback_batch2.head()
````

**Concatenate Two Feedback Datasets**

Both datasets have the same columns.

Therefore, we can stack them vertically.

````python
# Combine the two feedback batches row-wise
all_feedback = pd.concat(
    [df, feedback_batch2],
    ignore_index=True
)

# Display the combined dataset
all_feedback
````

**Important difference**

`concat()` is generally used when datasets have a similar structure and we want to stack them.

`merge()` is used when datasets have related information connected through a common key.

**PART F — FINAL DATA-CLEANING PIPELINE**

**Complete Cleaning Process**

Now let's put everything together.

````python
import pandas as pd

# Load the feedback dataset
df = pd.read_csv("customer_feedback_batch1.csv")

# Remove unnecessary spaces from text columns
df["City"] = df["City"].str.strip()
df["Response_Channel"] = df["Response_Channel"].str.strip()

# Standardize city names
df["City"] = df["City"].str.title()

# Standardize response channel names
df["Response_Channel"] = df["Response_Channel"].str.title()

# Fill missing numerical values with median
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Rating"] = df["Rating"].fillna(df["Rating"].median())

# Fill missing categorical values with mode
df["City"] = df["City"].fillna(df["City"].mode()[0])

# Fill missing text values with a meaningful label
df["Feedback"] = df["Feedback"].fillna("No Feedback")
df["Response_Channel"] = df["Response_Channel"].fillna("Unknown")

# Remove duplicate records
df = df.drop_duplicates()

# Display the cleaned dataset
df.head()
````

**Merge Cleaned Data with Customer Master**

````python
# Load customer master data
customer_master = pd.read_csv("customer_master.csv")

# Merge customer details with cleaned feedback
final_df = pd.merge(
    df,
    customer_master,
    on="Customer_ID",
    how="left"
)

# Display the final dataset
final_df.head()
````

**Validate the Final Dataset**

**Check missing values**

````python
# Check remaining missing values
final_df.isnull().sum()
````

**Check duplicates**

````python
# Count duplicate rows
final_df.duplicated().sum()
````

**Save the Cleaned Dataset**

Once the data has been cleaned and merged, save it for future analysis.

````python
# Save the final cleaned dataset as a CSV file
final_df.to_csv("cleaned_customer_feedback.csv", index=False)

# Display a confirmation message
print("Cleaned dataset saved successfully.")
````
