```python
# Rainfall Probability & Conditional Probability Analysis

import pandas as pd
import matplotlib.pyplot as plt

# 1. Load Dataset

df = pd.read_csv("weather_probability_dataset1.csv")

print("First 5 rows:")
display(df.head())

print("\nDataset Shape:", df.shape)

print("\nColumns:")
print(df.columns.to_list())

# 2. Check RainTomorrow Values

print("\nUnique values in RainTomorrow:")
print(df['RainTomorrow'].unique())

# 3. Basic Probability of Rain Tomorrow

print("\nRain Tomorrow Counts:")
print(df['RainTomorrow'].value_counts())

rainy_days = (df['RainTomorrow'] == 'Yes').sum()
total_days = len(df)

probability = rainy_days / total_days

print("\n--- Basic Probability ---")
print("Rainy Days:", rainy_days)
print("Total Days:", total_days)
print("Probability of Rain Tomorrow:", round(probability * 100, 2), "%")

# 4. Conditional Probability: P(RainTomorrow | RainToday)

rain_today_df = df[df['RainToday'] == 'Yes']

print("\nNumber of days when it rained today:", len(rain_today_df))

rain_tomorrow_after_rain_today = (rain_today_df['RainTomorrow'] == 'Yes').sum()

print("Number of those days when it also rained tomorrow:",rain_tomorrow_after_rain_today)

conditional_probability = (rain_tomorrow_after_rain_today /len(rain_today_df))

print("\nP(RainTomorrow | RainToday) =", round(conditional_probability * 100, 2), "%")

# 5. Conditional Probability Using Pandas mean()

conditional_probability1 = ((df[df['RainToday'] == 'Yes']['RainTomorrow'] == 'Yes').mean())

print("\nUsing Pandas mean():", round(conditional_probability1 * 100, 2), "%" )

# 6. Conditional Probability: P(RainTomorrow | High Humidity)

print("\nHighHumidity Counts:")
print(df['HighHumidity'].value_counts())

probability_high_humidity = ((df[df['HighHumidity'] == 'Yes']['RainTomorrow'] == 'Yes').mean())

print("\nP(RainTomorrow | High Humidity) =", round(probability_high_humidity * 100, 2),"%")

# 7. Conditional Probability: P(RainTomorrow | Low Pressure)

print("\nLowPressure Counts:")
print(df['LowPressure'].value_counts())

probability_low_pressure = (( df[df['LowPressure'] == 'Yes']['RainTomorrow'] == 'Yes').mean())

print("\nP(RainTomorrow | Low Pressure) =", round(probability_low_pressure * 100, 2),"%")


# 8. Conditional Probability: P(RainTomorrow | High Cloud Cover)

print("\nHighCloudCover Counts:")
print(df['HighCloudCover'].value_counts())

probability_high_cloud = (
    (
        df[df['HighCloudCover'] == 'Yes']['RainTomorrow']
        == 'Yes'
    ).mean()
)

print(
    "\nP(RainTomorrow | High Cloud Cover) =",
    round(probability_high_cloud * 100, 2),
    "%"
)

# 9. Compare All Probabilities

results = {
    'Condition': [
        'Overall',
        'Rain Today',
        'High Humidity',
        'Low Pressure',
        'High Cloud Cover'
    ],

    'Probability': [
        (df['RainTomorrow'] == 'Yes').mean(),

        (
            df[df['RainToday'] == 'Yes']['RainTomorrow']
            == 'Yes'
        ).mean(),

        (
            df[df['HighHumidity'] == 'Yes']['RainTomorrow']
            == 'Yes'
        ).mean(),

        (
            df[df['LowPressure'] == 'Yes']['RainTomorrow']
            == 'Yes'
        ).mean(),

        (
            df[df['HighCloudCover'] == 'Yes']['RainTomorrow']
            == 'Yes'
        ).mean()
    ]
}

results_df = pd.DataFrame(results)



# 10. Convert Probability to Percentage

results_df['Probability (%)'] = (
    results_df['Probability'] * 100
)

print("\n======================================")
print("Conditional Probability Results")
print("======================================")
display(results_df)


# 11. Visualization

plt.figure(figsize=(10, 5))

plt.bar(
    results_df['Condition'],
    results_df['Probability (%)']
)

plt.xlabel("Weather Condition")
plt.ylabel("Probability of Rain Tomorrow (%)")
plt.title("Conditional Probability of Rainfall")

plt.xticks(rotation=15)

plt.tight_layout()
plt.show()
```
