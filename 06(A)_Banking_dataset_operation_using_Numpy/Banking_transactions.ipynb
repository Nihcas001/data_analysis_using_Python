```python
# Step 1 — Import NumPy
import numpy as np
```

Column 0 → Customer ID

Column 1 → Balance

Column 2 → Deposit

Column 3 → Withdrawal

Column 4 → EMI

Column 5 → Transactions

```python
# Step 2 — Create the Dataset
data = np.array([
    [1001, 50000, 12000, 5000, 3000, 8],
    [1002, 72000, 15000, 7000, 4500, 12],
    [1003, 35000, 8000, 3000, 2500, 6],
    [1004, 91000, 20000, 9000, 6000, 15],
    [1005, 45000, 10000, 4000, 3000, 9],
    [1006, 68000, 14000, 6500, 4000, 11],
    [1007, 29000, 7000, 2500, 2000, 5],
    [1008, 83000, 18000, 8500, 5000, 14],
    [1009, 57000, 11000, 5500, 3500, 10],
    [1010, 102000, 25000, 12000, 7000, 18],
    [1011, 39000, 9000, 3500, 2500, 7],
    [1012, 76000, 16000, 7500, 4500, 13],
    [1013, 48000, 10500, 4500, 3000, 8],
    [1014, 95000, 22000, 10000, 6500, 16],
    [1015, 62000, 13000, 6000, 4000, 11]
])

print(data)
```

```python
# Step 3 — Understand the Array
# Question 1: What is the number of dimensions?
print(data.ndim)
```

Because our data has: Rows × Columns

So it is a 2-dimensional array.

```python
# Question 2: What is the shape?
print(data.shape)
```

```python
# Question 3: How many elements are present?
print(data.size)
```

Because: 15 × 6 = 90

```python
# Question 4: What is the data type?
print(data.dtype)
```

**Step 4 — Basic Indexing**

```python
# Question 5: Display the first customer's complete record.
print(data[0])
```

So:

data[0] → first row

data[1] → second row

data[2] → third row

```python
# Question 6: Display the first customer's balance.
# Balance is column 1.

print(data[0, 1]) # data[row, column]
```

```python
# Question 7: Display customer ID of the 10th customer.
print(data[9, 0])
```

**8. Negative Indexing**

Negative indexing starts from the end.

-1 → last

-2 → second last

-3 → third last

```python
# Question 8: Display the last customer's record.

print(data[-1])
```

```python
# Question 9: Display the last customer's balance.
print(data[-1, 1])
```

**9. Slicing**

Slicing is extremely important.

General syntax: `array[start:stop]`

Remember: stop is excluded.

```python
# Question 10: Display the first five customers.
print(data[:5]) # This gives: Rows 0,1,2,3,4
```

```python
# Question 11: Display customers 6 to 10.
print(data[5:10])
```

```python
# Question 12: Display the last five customers.
print(data[-5:])
```

**10. Column Slicing**

```python
# Question 13: Extract all customer IDs.
customer_ids = data[:, 0]  # colon(:) means All rows, column 0.
print(customer_ids)
```

```python
# Question 14: Extract all account balances.
balances = data[:, 1]
print(balances)
```

```python
# Question 15: Extract all deposits.

deposits = data[:, 2]
print(deposits)
```

```python
# Question 16: Extract all withdrawals.

withdrawals = data[:, 3]
print(withdrawals)
```

**11. Extract Multiple Columns**

```python
# Question 17: Extract Balance, Deposit and Withdrawal columns.

financial_data = data[:, 1:4]
print(financial_data)

# Why 1:4?
# 1 → Balance
# 2 → Deposit
# 3 → Withdrawal
```

**12. Arithmetic Operations**

Suppose the bank wants to calculate the new balance after deposit and withdrawal.

Formula: `New Balance = Balance + Deposit - Withdrawal`

```python
# Question 18: Calculate new balance for every customer.

new_balance = data[:, 1] + data[:, 2] - data[:, 3]
print(new_balance)

# NumPy performs the calculation element by element.
```

**13. Calculate EMI-adjusted Balance**


```python
# Question 19: Calculate balance after EMI deduction.
# Formula: Remaining Balance = Balance - EMI

remaining_balance = data[:, 1] - data[:, 4]
print(remaining_balance)
```

**14. Scalar Arithmetic**

Suppose the bank wants to give 5% interest on current balances.

```python
# Question 20: Calculate 5% interest.

interest = data[:, 1] * 0.05
print(interest)
```

```python
# Question 21: Calculate balance after adding 5% interest.

balance_with_interest = data[:, 1] * 1.05
print(balance_with_interest)
```

**15. Array-to-Array Operations**

Suppose the bank wants to compare deposits and withdrawals.

```python
# Question 22: Calculate net transaction amount.
net_transaction = data[:, 2] - data[:, 3] # Deposit - Withdrawal
print(net_transaction)
```

**16. Aggregation Operations**

Now we can calculate overall banking statistics.

```python
# Question 23: What is the total balance of all customers?
total_balance = np.sum(data[:, 1])
print(total_balance)
```

```python
# Question 24: What is the average account balance?
average_balance = np.mean(data[:, 1])
print(average_balance)
```

```python
# Question 25: Find the highest account balance.
maximum_balance = np.max(data[:, 1])
print(maximum_balance)
```

```python
# Question 26: Find the lowest account balance.
minimum_balance = np.min(data[:, 1])
print(minimum_balance)
```

**17. Finding Position of Maximum/Minimum**

```python
# Question 27: Which customer's balance is highest?
index = np.argmax(data[:, 1])

print("Index:", index)
print("Customer ID:", data[index, 0])
print("Balance:", data[index, 1])

# np.argmax() returns the position of the maximum value.
```

```python
# Similarly for min
index = np.argmin(data[:, 1])
print("Customer ID:", data[index, 0])
```

**18. Boolean Indexing**

```python
# Question 28: Find customers whose balance is greater than ₹70,000.

high_balance = data[data[:, 1] > 70000]
print(high_balance)
```
