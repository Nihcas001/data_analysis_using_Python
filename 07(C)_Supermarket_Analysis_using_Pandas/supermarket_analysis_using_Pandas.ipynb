````python
# Load the supermarket sales dataset
df = pd.read_csv("supermarket_sales_analysis.csv")

# Display the dataset
df
````

**Part 2 — What is Grouping?**

Suppose the supermarket manager asks:

>How much total sales did each category generate?

Our dataset contains many transactions:
```
Grocery
Beverages
Grocery
Dairy
Beverages
Grocery
...
```
Instead of manually calculating each category, Pandas can group the records according to Category.

The main function is:

`groupby()`

**Step 5: Basic groupby()**

````python
# Group sales records according to product category
category_group = df.groupby("Category")

# Display the grouped object
print(category_group)
````

The result is a GroupBy object.

It does not perform a calculation yet.

We need to tell Pandas what calculation we want.

**Part 3 — Aggregation
What is Aggregation?**

Aggregation means:

> Summarizing multiple records into a single value.

Common aggregation functions are:
```
Function	Meaning
sum()	    Total
mean()	    Average
count()	    Number of records
min()	    Minimum
max()	    Maximum
median()	Middle value
```

**Step 6: Total Sales by Category**

This is the most important operation for this problem.

````python
# Calculate total sales for each category
category_sales = df.groupby("Category")["Sales_Amount"].sum()

# Display total sales by category
category_sales
````

**Conceptually**:

```
Category
    ↓
Group records
    ↓
Sales_Amount
    ↓
Calculate SUM
```

**Step 7: Total Quantity Sold by Category**

We can also calculate the total number of units sold.

````python
# Calculate total quantity sold in each category
category_quantity = df.groupby("Category")["Quantity"].sum()

# Display the result
category_quantity
````

**Step 8: Average Sales by Category**

Instead of total sales, suppose the manager wants the average transaction value.

````python
# Calculate average sales amount for each category
average_sales = df.groupby("Category")["Sales_Amount"].mean()

# Display average sales
average_sales
````

**Step 9: Number of Transactions per Category**

We can use *count()*.

````python
# Count the number of sales transactions in each category
category_count = df.groupby("Category")["Sale_ID"].count()

# Display transaction count
category_count
````

**Step 10: Minimum Sales by Category**

````python
# Find the minimum transaction value for each category
minimum_sales = df.groupby("Category")["Sales_Amount"].min()

# Display minimum sales
minimum_sales
````

**Step 11: Maximum Sales by Category**

````python
# Find the maximum transaction value for each category
maximum_sales = df.groupby("Category")["Sales_Amount"].max()

# Display maximum sales
maximum_sales
````

**Part 4 — Multiple Aggregations**

A very useful Pandas feature is performing multiple aggregations at once.

Suppose the supermarket manager wants:

*   Total sales
*   Average sales
*   Minimum sales
*   Maximum sales

for every category.

We can use **.agg().**

**Step 12: Multiple Aggregations**

````python
# Calculate multiple sales statistics for each category
category_summary = df.groupby("Category")["Sales_Amount"].agg(
    ["sum", "mean", "min", "max"]
)

# Display the category-wise summary
category_summary
````

**Part 5 — Named Aggregations**

We can give meaningful names to the output columns.

````python
# Create a readable category-wise sales summary
category_summary = df.groupby("Category").agg(
    Total_Sales=("Sales_Amount", "sum"),
    Average_Sales=("Sales_Amount", "mean"),
    Total_Quantity=("Quantity", "sum"),
    Transactions=("Sale_ID", "count")
)

# Display the summary
category_summary
````

**Part 6 — Grouping by Multiple Columns**

We can group data using more than one column.

For example:

> Find total sales for every category in every store.

We need:

`Store + Category`

**Step 13: Group by Store and Category**

````python
# Calculate total sales for each store-category combination
store_category_sales = df.groupby(
    ["Store", "Category"]
)["Sales_Amount"].sum()

# Display the result
store_category_sales
````

**Step 14: Convert Grouped Result into a DataFrame**

Sometimes we want the result in normal tabular form.

Use:

**`reset_index()`**

````python
# Convert the grouped result back into a DataFrame
store_category_sales = df.groupby(
    ["Store", "Category"]
)["Sales_Amount"].sum().reset_index()

# Display the result
store_category_sales
````

**Part 7 — Sorting Grouped Results**

Suppose we want to find which category generated the highest sales.

````python
# Calculate total sales by category
category_sales = df.groupby("Category")["Sales_Amount"].sum()

# Sort categories from highest to lowest sales
category_sales = category_sales.sort_values(ascending=False)

# Display the sorted result
category_sales
````

**Step 15: Find the Highest-Selling Category**

idxmax() returns the index/label corresponding to the maximum value.

````python
# Find the category with the highest total sales
highest_category = category_sales.idxmax()

# Display the category name
print("Highest-selling category:", highest_category)
````

**Part 8 — What is a Pivot Table?**

A pivot table is a powerful way of summarizing data from different dimensions.

Suppose the supermarket manager asks:

> Show me sales for each category across different stores.

We can create:

| Category      | Chandigarh | Delhi | Mumbai | Pune |
| ------------- | ---------: | ----: | -----: | ---: |
| Beverages     |        ... |   ... |    ... |  ... |
| Dairy         |        ... |   ... |    ... |  ... |
| Grocery       |        ... |   ... |    ... |  ... |
| Personal Care |        ... |   ... |    ... |  ... |


This is exactly what a pivot table is useful for.

**Step 16 — Basic Pivot Table**

The Pandas function is:

`pd.pivot_table()`

````python
# Create a pivot table showing category-wise sales by store
pivot_sales = pd.pivot_table(
    df,                    # The source DataFrame
    values="Sales_Amount", # The numerical column we want to analyze.
    index="Category",      # The rows of the pivot table.
    columns="Store",       # The columns of the pivot table.
    aggfunc="sum"          # The calculation to perform.
)

# Display the pivot table
pivot_sales
````

**Step 17 — Pivot Table Using Average**

Instead of total sales, we can calculate average sales.

````python
# Create a pivot table showing average sales by category and store
pivot_average = pd.pivot_table(
    df,
    values="Sales_Amount",
    index="Category",
    columns="Store",
    aggfunc="mean"
)

# Display the pivot table
pivot_average
````

**Step 18 — Pivot Table Using Quantity**

We can analyze quantity instead of sales.

````python
# Create a pivot table showing total quantity sold by category and store
pivot_quantity = pd.pivot_table(
    df,
    values="Quantity",
    index="Category",
    columns="Store",
    aggfunc="sum"
)

# Display the pivot table
pivot_quantity
````

**Step 19 — Add Total Row and Column**

A pivot table can automatically calculate totals using **margins=True.**

````python
# Create a pivot table with category-store totals
pivot_total = pd.pivot_table(
    df,
    values="Sales_Amount",
    index="Category",
    columns="Store",
    aggfunc="sum",
    margins=True
)

# Display category, store, and grand totals
pivot_total
````

**Step 20 — Replace Missing Pivot Values**

Some category-store combinations may not exist in the dataset.

These can appear as **NaN**.

We can replace them with zero.

````python
# Replace missing pivot table values with zero
pivot_sales = pivot_sales.fillna(0)

# Display the cleaned pivot table
pivot_sales
````

**Part 9 — Grouping by Month**

Since our dataset contains dates, we can also perform monthly sales analysis.

First, make sure the Date column is a datetime column.

````python
# Convert the Date column into Pandas datetime format
df["Date"] = pd.to_datetime(df["Date"])
````

**Step 21: Extract Month**

````python
# Extract the month from each sale date
df["Month"] = df["Date"].dt.month

# Display the first few records
df[["Date", "Month"]].head()
````

**Step 22: Monthly Sales**

````python
# Calculate total sales for each month
monthly_sales = df.groupby("Month")["Sales_Amount"].sum()

# Display monthly sales
monthly_sales
````

**Step 23 — Month and Category Analysis**

We can combine multiple dimensions.

````python
# Calculate monthly sales for each category
monthly_category_sales = df.groupby(
    ["Month", "Category"]
)["Sales_Amount"].sum().reset_index()

# Display the result
monthly_category_sales
````

**Step 24 — Monthly Category Pivot Table**

This can be represented more clearly using a pivot table.

````python
# Create a pivot table for monthly category-wise sales
monthly_pivot = pd.pivot_table(
    df,
    values="Sales_Amount",
    index="Month",
    columns="Category",
    aggfunc="sum",
    fill_value=0
)

# Display the monthly sales analysis
monthly_pivot
````

**Part 10 — Complete Category-Wise Analysis**

Now let's create one complete analysis that a supermarket manager could actually use.

This is a strong example of grouping + aggregation + sorting.

````python
# Generate a complete category-wise sales summary
category_analysis = df.groupby("Category").agg(
    Total_Sales=("Sales_Amount", "sum"),
    Average_Sale=("Sales_Amount", "mean"),
    Total_Quantity=("Quantity", "sum"),
    Number_of_Transactions=("Sale_ID", "count"),
    Maximum_Sale=("Sales_Amount", "max"),
    Minimum_Sale=("Sales_Amount", "min")
)

# Sort categories by total sales
category_analysis = category_analysis.sort_values(
    "Total_Sales",
    ascending=False
)

# Display the final category analysis
category_analysis
````

**Part 11 — Complete Pivot Analysis**

````python
# Create a category-store sales pivot table
category_store_pivot = pd.pivot_table(
    df,
    values="Sales_Amount",
    index="Category",
    columns="Store",
    aggfunc="sum",
    fill_value=0,
    margins=True
)

# Display the final pivot analysis
category_store_pivot
````

This lets the supermarket answer questions such as:

* Which category has the highest sales?
* Which store generates the highest sales?
* How much does each category contribute to each store?
* Which category performs poorly at a particular store?
* What is the overall sales total?
