```python
# Step 1 — Import libraries
import pandas as pd
import numpy as np
from scipy import stats
```

```python
# Step 2 — Load the dataset
df = pd.read_csv("manufacturing_quality_data.csv")
```

```python
# Display df
df
```

```python
# display top 5 rows
df.head()
```

```python
# Step 3 — Inspect the dataset
len(df)
```

```python
# Check the quality column
df["Quality_Score"]
```

```python
# Step 4 — Calculate the sample mean

sample_mean = df["Quality_Score"].mean()

print("Sample Mean:", sample_mean)
```

**The sample mean tells us the average quality score of products made using the new process.**

```python
# Step 5 — Define the known values

mu_0 = 78       # Historical population mean
sigma = 5       # Known population standard deviation
n = len(df)     # Sample size
alpha = 0.05    # Significance level
```

**Step 6 — Calculate the Z-score manually**
Screenshot 2026-08-19 172916.png


```python
# Apply the formula of Z-test
z = (sample_mean - mu_0) / (sigma / np.sqrt(n))
print("Z-score:", round(z, 2))
```

**Step 7 — Find the critical value**

Because this is a right-tailed test with α = 0.05

```python
# Step 7
critical_value = stats.norm.ppf(1 - alpha)

print("Critical Value:", round(critical_value, 3))
```

```python
# Step 8 — Make the decision

if z > critical_value:
    print("Reject the Null Hypothesis")
else:
    print("Fail to Reject the Null Hypothesis")
```

**Step 9 — Calculate the p-value**

The p-value tells us how strong the evidence is against H₀.

The p-value will be extremely small.

For a right-tailed test:

```python
p_value = 1 - stats.norm.cdf(z)

print("P-value:", round(p_value, 2))
```

```python
# Make the decision

if p_value < alpha:
    print("Reject the Null Hypothesis")
else:
    print("Fail to Reject the Null Hypothesis")
```

Because the p-value is much smaller than 0.05:

**Reject H₀.**

**Step 10 — Perform the complete Z-test using SciPy**

We can also calculate the Z-test directly.

```python
# Step 10 - calculate z score directly

z_stat = (sample_mean - mu_0) / (sigma / np.sqrt(n))

p_value = 1 - stats.norm.cdf(z_stat)

print("Sample Mean:", sample_mean)
print("Z-statistic:", round(z_stat, 2))
print("P-value:", round(p_value,2))
```
